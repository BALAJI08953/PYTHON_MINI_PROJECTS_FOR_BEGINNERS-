```python
#keyword arugments 
def hello(name,age=20):
    print('my name is {} and age is {}'.format(name,age))
hello('Bhagirath')

#no assingment in the function is positional argument
##assingment is yes then its a keyword argument 
##to convert back you dont need to assign any problem

```

```python
#new way of definition of positional and keyword arugments
def hello_2(*args,**kwargs):
    print(args)
    print(kwargs)
hello_2("BHAGIRATH",age=20,dob=2006)
```

```python
lis=['BHAGIRATH']
dist={'age': 20, 'dob': 2006}
hello_2(*lis,**dist)
```

```python
lst2=[1,2,3,4,5,6,7,8,9]
def even_sum(lst2):
    even_sums=0
    odd_sums=0
    for i in lst2:
        if i%2==0:
            even_sums=even_sums+i
        else:
            odd_sums=odd_sums+i
    return even_sums,odd_sums
```

```python
even_sum(lst2)
```

```python
#map functions in python 
def checksum(num):
    if num%2==0:
        return "the num is even"
    else:
        return "the num is odd"
```

```python
checksum(25)
```

```python
col=[1,2,4,5,6,8,9,5,6,3,2,5,78,9,45,6,23,0,3,4,8]
#usage of map function
list(map(checksum,col))
```
