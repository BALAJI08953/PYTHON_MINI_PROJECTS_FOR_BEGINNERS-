```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def startengine(self):
        pass

# Derived class 1
class Bus(Vehicle):
    def startengine(self):
        return "Bus engine has been started"

# Derived class 2
class Truck(Vehicle):
    def startengine(self):
        return "Truck engine has been started"

def start_vehicle(vehicle_obj):
    print(f"The {vehicle_obj.startengine()}")

# Creating objects and calling the function
bus = Bus()
truck = Truck()

start_vehicle(bus)
start_vehicle(truck)
```
